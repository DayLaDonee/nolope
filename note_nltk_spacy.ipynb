

# 1. NLTK (Natural Language Toolkit)

| **Chức năng** | **Mô tả** | **Ví dụ** | **Output** |
|---------------|-----------|-----------|------------|
| `word_tokenize(text)` | Tách từ (tokenize từ). | `nltk.word_tokenize("Hello world!")` | `['Hello', 'world', '!']` |
| `sent_tokenize(text)` | Tách câu (sentence tokenize). | `nltk.sent_tokenize("Hello world! How are you?")` | `['Hello world!', 'How are you?']` |
| `pos_tag(tokens)` | Gán nhãn từ loại (Part-of-Speech tagging). | `nltk.pos_tag(["Hello", "world"])` | `[('Hello', 'NNP'), ('world', 'NN')]` |
| `ne_chunk(tree)` | Phân tích thực thể tên (Named Entity Recognition). | `nltk.ne_chunk(nltk.pos_tag(nltk.word_tokenize("Barack Obama was born in Hawaii")))` | `(S (PERSON Barack/NNP Obama/NNP) was/VBD born/VBN in/IN Hawaii/NNP)` |
| `FreqDist(tokens)` | Thống kê tần suất xuất hiện từ. | `FreqDist(['dog', 'dog', 'cat'])` | `{'dog': 2, 'cat': 1}` |
| `stopwords.words('english')` | Danh sách stopwords tiếng Anh. | `from nltk.corpus import stopwords; stopwords.words('english')[:5]` | `['i', 'me', 'my', 'myself', 'we']` |
| `nltk.corpus.brown.words()` | Truy cập Brown Corpus. | `nltk.corpus.brown.words()[:5]` | `['The', 'Fulton', 'County', 'Grand', 'Jury']` |
| `nltk.Tree` | Tạo cây cú pháp. | `nltk.Tree('S', ['NP', 'VP'])` | `(S NP VP)` |
| `PorterStemmer().stem(word)` | Rút gọn từ gốc (stemming). | `PorterStemmer().stem("running")` | `'run'` |
| `WordNetLemmatizer().lemmatize(word)` | Lemmatization (dạng chuẩn từ). | `WordNetLemmatizer().lemmatize("running", pos='v')` | `'run'` |

# 2. spaCy

| **Chức năng** | **Mô tả** | **Ví dụ** | **Output** |
|---------------|-----------|-----------|------------|
| `nlp(text)` | Phân tích văn bản thành Doc object. | `doc = nlp("Hello world!")` | `Doc object` |
| `doc.text` | Lấy văn bản gốc. | `doc.text` | `'Hello world!'` |
| `doc.ents` | Danh sách các thực thể tên. | `[ent.text for ent in nlp("Barack Obama was born in Hawaii.").ents]` | `['Barack Obama', 'Hawaii']` |
| `doc.sents` | Tách câu. | `[sent.text for sent in nlp("Hello world! How are you?").sents]` | `['Hello world!', 'How are you?']` |
| `[token.text for token in doc]` | Tách token từ văn bản. | `[token.text for token in nlp("Hello world!")]` | `['Hello', 'world', '!']` |
| `token.pos_` | Gán nhãn từ loại. | `[token.pos_ for token in nlp("Hello world!")]` | `['INTJ', 'NOUN', 'PUNCT']` |
| `token.dep_` | Phân tích quan hệ cú pháp. | `[token.dep_ for token in nlp("Hello world!")]` | `['discourse', 'ROOT', 'punct']` |
| `token.lemma_` | Lemmatization từ. | `[token.lemma_ for token in nlp("running runs ran")]` | `['run', 'run', 'run']` |
| `token.is_stop` | Kiểm tra stopword. | `[token.is_stop for token in nlp("This is a test")]` | `[True, True, False, False]` |
| `spacy.matcher.Matcher` | Khớp mẫu token nâng cao. | `Matcher(nlp.vocab)` | `Matcher object` |
| `doc.similarity(doc2)` | Tính độ tương đồng giữa 2 văn bản Doc. | `doc1.similarity(doc2)` | `0.79` (giá trị cosine) |
| `token.similarity(token2)` | Tính độ tương đồng giữa 2 token (từ đơn). | `token1.similarity(token2)` | `0.46` |
| `Matcher(nlp.vocab)` | Tạo bộ lọc mẫu dựa trên POS tags. | `Matcher.add("VerbPhrase", [pattern])` | Tìm các cụm từ thỏa mẫu POS |

# 3.Gensim

| **Chức năng** | **Mô tả** | **Ví dụ** | **Output** |
|---------------|-----------|-----------|------------|
| `gensim.utils.simple_preprocess(text)` | Tiền xử lý văn bản: lowercase, bỏ dấu câu. | `simple_preprocess("Hello World, NLP!")` | `['hello', 'world', 'nlp']` |
| `gensim.corpora.Dictionary(list_of_docs)` | Tạo từ điển (mapping từ → id). | `Dictionary([['hello', 'world'], ['world', 'machine', 'learning']])` | `{hello: 0, world: 1, machine: 2, learning: 3}` |
| `dictionary.doc2bow(doc)` | Chuyển văn bản thành bag-of-words vector. | `dictionary.doc2bow(['hello', 'machine'])` | `[(0, 1), (2, 1)]` |
| `models.TfidfModel(corpus)` | Huấn luyện mô hình TF-IDF trên corpus. | `TfidfModel(bow_corpus)` | Mô hình TF-IDF |
| `tfidf_model[bow_vector]` | Biến đổi vector BOW thành vector TF-IDF. | `tfidf_model[dictionary.doc2bow(['hello', 'world'])]` | `[(0, 0.7071), (1, 0.7071)]` |
| `models.LdaModel(corpus, num_topics, id2word)` | Huấn luyện mô hình LDA (Topic Modeling). | `LdaModel(corpus=bow_corpus, id2word=dictionary, num_topics=2)` | Mô hình chủ đề (topics) |
| `lda_model.print_topics()` | In ra danh sách các chủ đề học được. | `lda_model.print_topics(num_words=4)` | `[('0', '0.5*"world" + 0.5*"machine"'), ('1', '0.5*"hello" + 0.5*"learning"')]` |
| `KeyedVectors.load_word2vec_format(path)` | Load mô hình Word2Vec pretrained. | `KeyedVectors.load_word2vec_format('GoogleNews-vectors-negative300.bin', binary=True)` | Word2Vec model |
| `word_vectors.most_similar('king')` | Tìm từ gần nhất với từ cho trước. | `model.most_similar('king')` | `[(queen, 0.78), (prince, 0.76), ...]` |

# 1.Giới thiệu chung

| Thư viện | Mô tả chi tiết |
|:---------|:----------------|
| **NLTK** | Thư viện NLP truyền thống, hữu ích cho bài tập học thuật, tokenization, POS tagging, stemming, stopword removal. |
| **spaCy** | Thư viện NLP hiện đại, nhanh chóng, dùng cho dự án thực tế, mạnh về NER, POS, dependency parsing, similarity. |
| **Gensim** | Thư viện chuyên dùng embeddings (Word2Vec, Doc2Vec), topic modeling (LDA/LSA), xử lý tập dữ liệu lớn. |

# 2.Chức năng chi tiết

| Thư viện | Chức năng | Mô tả chi tiết | Ví dụ | Output |
|:---------|:----------|:-------------|:------|:-------|
| **NLTK** | Tokenization | Tách từ/câu | `word_tokenize('Hello world!')` | `['Hello', 'world', '!']` |
| **NLTK** | POS Tagging | Gán nhãn từ loại | `pos_tag(['Hello', 'world'])` | `[('Hello', 'NNP'), ('world', 'NN')]` |
| **NLTK** | Named Entity Recognition | Phát hiện thực thể | `ne_chunk(pos_tag(['Barack', 'Obama']))` | Tree(PERSON Barack/NNP Obama/NNP) |
| **NLTK** | Stopword Removal | Loại bỏ stopwords | `stopwords.words('english')` | Danh sách stopwords |
| **NLTK** | Stemming | Lấy gốc từ | `PorterStemmer().stem('running')` | `'run'` |
| **NLTK** | Lemmatization | Chứng hoá từ | `WordNetLemmatizer().lemmatize('better', pos='a')` | `'good'` |
| **NLTK** | FreqDist | Phân bố tần suất từ | `FreqDist(['dog', 'dog', 'cat'])` | `{'dog': 2, 'cat': 1}` |
| **spaCy** | Tokenization | Tách từ và câu | `for token in nlp('Hello world!')` | `Hello`, `world`, `!` |
| **spaCy** | POS Tagging | Gán nhãn POS chuẩn | `token.pos_` | NOUN, VERB, etc. |
| **spaCy** | Named Entity Recognition | Phát hiện thực thể | `ent.text`, `ent.label_` | PERSON, ORG, GPE,... |
| **spaCy** | Dependency Parsing | Phân tích quan hệ câu | `token.dep_`, `token.head.text` | nsubj, dobj, root,... |
| **spaCy** | Similarity Computation | Tính độ tương đồng văn bản | `doc1.similarity(doc2)` | float (0.0 - 1.0) |
| **spaCy** | Lemmatization | Chuẩn hoá từ | `token.lemma_` | gốc từ ("running" → "run") |
| **Gensim** | Word2Vec Training | Huấn luyện embeddings | `Word2Vec(sentences)` | Mô hình embeddings |
| **Gensim** | TF-IDF Representation | Biểu diễn tf-idf | `TfidfModel(bow_corpus)` | Ma trận tf-idf |
| **Gensim** | Topic Modeling (LDA) | Rut trích chủ đề | `LdaModel(corpus, num_topics=10)` | Các chủ đề |
